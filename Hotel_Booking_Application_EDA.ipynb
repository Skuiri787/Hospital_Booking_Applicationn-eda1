## 4. Data Cleaning

```python
# Counting Missing Values

df.isnull().sum().sort_values(ascending=False)
```

```python
# Showing only missing value columns and their percentage
missing_df = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Percentage': round((df.isnull().sum()/len(df))*100,2)
})

missing_df[missing_df['Missing Values'] > 0].sort_values(by='Percentage', ascending=False)
```

```python
# Handling and Imputing Missing values

# Fill missing company IDs with 0 (No Company)
df['company'] = df['company'].fillna(0)

# Fill missing agent IDs with 0 (No Agent)
df['agent'] = df['agent'].fillna(0)

# Fill missing country values with the most frequent country
df['country'] = df['country'].fillna(df['country'].mode()[0])

# Fill missing children values with the median
df['children'] = df['children'].fillna(df['children'].median())
```

```python
# Checking duplicates

df.duplicated().sum()
```

```python
duplicates = df[df.duplicated(keep=False)]
duplicates.head(10)
```

```python
# Remove duplicate records
df = df.drop_duplicates()
```

```python
# Verify duplicates have been removed
df.duplicated().sum()
```

```python
df.shape
```

```python
# Converting to proper data type

df['reservation_status_date'] = pd.to_datetime(
    df['reservation_status_date'],
    format='mixed'
)
```

```python
df.info()
```

```python
# Saving the cleaned-processed dataset
df.to_csv("hotel_bookings_cleaned.csv", index=False)
```

### Summary

- Missing values were identified in four columns: `company`, `agent`, `country`, and `children`.
- Missing values in the `company` and `agent` columns were replaced with `0`, representing bookings made without a company or travel agent.
- Missing values in the `country` column were imputed using the most frequent country (mode), while missing values in the `children` column were filled using the median.
- A total of **32,013 duplicate records** were identified and removed, reducing the dataset size from **119,390** to **87,377** unique records.
- The `reservation_status_date` column was successfully converted from **object** to **datetime** data type to support time-based analysis.
- After completing the cleaning and preprocessing steps, the cleaned dataset was saved as **`hotel_bookings_cleaned.csv`** for future analysis and to ensure reproducibility without repeating the data cleaning process.
- The final cleaned dataset contained no missing values in the previously affected columns and was fully prepared for further exploratory data analysis.

## 7. Time Series Analysis

```python
#1. How do hotel bookings vary over time?

# Create a proper arrival date column
df['arrival_date'] = pd.to_datetime(
    df['arrival_date_day_of_month'].astype(str) + ' ' +
    df['arrival_date_month'] + ' ' +
    df['arrival_date_year'].astype(str)
)

df[['arrival_date']].head()
```

```python
# Monthly booking trend
monthly_bookings = (
    df.groupby(df['arrival_date'].dt.to_period('M'))
      .size()
      .reset_index(name='Bookings')
)

monthly_bookings['arrival_date'] = monthly_bookings['arrival_date'].astype(str)

plt.figure(figsize=(12,5))

ax = sns.lineplot(
    data=monthly_bookings,
    x='arrival_date',
    y='Bookings',
    marker='o'
)

plt.title('Monthly Booking Trend', fontsize=14, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Number of Bookings')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Key Observation

- Hotel bookings exhibit a clear seasonal pattern across the observed period.
- Booking demand generally peaks during the **summer months (July–October)**, with the highest number of bookings recorded around **May–October 2017**.
- The lowest booking volumes are observed during the **winter months (November–January)**, indicating an off-season for hotel reservations.
- A recurring seasonal trend is visible across multiple years, suggesting that booking demand is influenced by holiday seasons and travel patterns.
- These trends can help hotel management optimize pricing strategies, staffing levels, and room inventory during peak and off-peak periods.

```python
#2 How do monthly bookings and cancellations change over time, and do cancellations follow the booking trend?

# Monthly Bookings vs Cancellations

monthly_summary = (
    df.groupby(df['arrival_date'].dt.to_period('M')).agg(
          Total_Bookings=('hotel', 'count'),
          Cancellations=('is_canceled', 'sum')
      ).reset_index()
)

monthly_summary['arrival_date'] = monthly_summary['arrival_date'].astype(str)

plt.figure(figsize=(12,6))

plt.plot(
    monthly_summary['arrival_date'],
    monthly_summary['Total_Bookings'],
    marker='o',
    linewidth=2,
    label='Total Bookings'
)

plt.plot(
    monthly_summary['arrival_date'],
    monthly_summary['Cancellations'],
    marker='o',
    linewidth=2,
    label='Cancellations'
)

plt.title('Monthly Bookings vs Cancellations', fontsize=14, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Number of Bookings')
plt.xticks(rotation=45)

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

### Key Observation

- Monthly bookings and cancellations exhibit a similar trend over the observed period, indicating that higher booking volumes are generally accompanied by higher cancellation volumes.
- Booking demand shows a gradual upward trend from 2015 to 2017, with noticeable peaks during the middle of each year.
- Cancellations also increase during high-demand periods but remain significantly lower than the total number of bookings throughout the timeline.
- Temporary declines in both bookings and cancellations are observed around the end and beginning of each year, suggesting seasonal fluctuations in hotel demand.
- Despite fluctuations in cancellations, the gap between total bookings and cancellations remains substantial across all months, indicating consistent booking retention over the study period.

```python
#3 How has the booking cancellation rate changed over time?

# Monthly Cancellation Rate (%)

monthly_rate = (
    df.groupby(df['arrival_date'].dt.to_period('M'))
      .agg(
          Total_Bookings=('hotel', 'count'),
          Cancellations=('is_canceled', 'sum')
      )
      .reset_index()
)

# Calculate cancellation rate
monthly_rate['Cancellation_Rate'] = (
    monthly_rate['Cancellations'] /
    monthly_rate['Total_Bookings']
) * 100

monthly_rate['arrival_date'] = monthly_rate['arrival_date'].astype(str)

plt.figure(figsize=(12,5))

plt.plot(
    monthly_rate['arrival_date'],
    monthly_rate['Cancellation_Rate'],
    marker='o',
    linewidth=2,
    color='crimson'
)

plt.title('Monthly Cancellation Rate Trend', fontsize=14, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Cancellation Rate (%)')

plt.xticks(monthly_rate.index[::2],
           monthly_rate['arrival_date'][::2],
           rotation=45)

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

### Key Observation

- The monthly cancellation rate shows an overall increasing trend throughout the study period, rising from approximately **15–20%** in late 2015 to around **35–37%** by mid-2017.
- The lowest cancellation rate is observed around **November 2015** (approximately **15%**), while the highest rate occurs in **August 2017** (approximately **37%**).
- Although the cancellation rate fluctuates across months, it remains consistently above **25%** during most of 2016 and continues to increase in 2017.
- The rising cancellation rate indicates that a growing proportion of bookings were being canceled over time, highlighting the need to investigate the underlying factors contributing to cancellations.
- Monitoring cancellation rates can help hotel management refine booking policies, pricing strategies, and customer retention initiatives to minimize revenue loss.

```python
#4 How does the Average Daily Rate (ADR) vary across different months of the year?

monthly_adr = (
    df.groupby('arrival_date_month', observed=False)['adr']
      .mean()
      .reindex(month_order)
      .reset_index()
)

plt.figure(figsize=(12,5))

ax = sns.barplot(
    data=monthly_adr,
    x='arrival_date_month',
    y='adr',
    palette='tab10'
)

for container in ax.containers:
    ax.bar_label(container, fmt='%.1f')

plt.title('Average Daily Rate (ADR) by Month', fontsize=14, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Average ADR')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Key Observation

- The **Average Daily Rate (ADR)** exhibits a clear seasonal pattern throughout the year.
- ADR steadily increases from **January (70.1)** and reaches its highest value in **August (150.9)**, indicating peak room pricing during the high-demand travel season.
- After August, the average room rate declines gradually, dropping to around **72.8** in **November** before showing a slight recovery in **December (81.4)**.
- The substantial variation in ADR across months suggests that the hotel adopts a **seasonal pricing strategy**, charging higher room rates during periods of increased demand and lowering prices during off-peak months.
- These insights can help hotel management optimize pricing policies, forecast revenue, and plan promotional campaigns during low-demand periods.

```python
#1	How does the lead time affect the likelihood of booking cancellations?

# Create lead time bins
lead_bins = [0, 30, 60, 90, 180, 365, df["lead_time"].max()]
lead_labels = [
    "0-30",
    "31-60",
    "61-90",
    "91-180",
    "181-365",
    "365+"
]

df["lead_time_group"] = pd.cut(
    df["lead_time"],
    bins=lead_bins,
    labels=lead_labels,
    include_lowest=True
)

# Calculate cancellation rate
lead_cancel = (
    df.groupby("lead_time_group", observed=True)["is_canceled"]
      .mean()
      .mul(100)
)

plt.figure(figsize=(9,5))

plt.plot(
    lead_cancel.index.astype(str),
    lead_cancel.values,
    marker="o",
    linewidth=2
)

# Data labels
for x, y in zip(lead_cancel.index.astype(str), lead_cancel.values):
    plt.text(x, y + 1, f"{y:.1f}%", ha="center", fontsize=10)

plt.title("Cancellation Rate by Lead Time",fontsize=14, fontweight='bold')
plt.xlabel("Lead Time (Days)")
plt.ylabel("Cancellation Rate (%)")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

### Key Observation
- Booking cancellations increase steadily as lead time increases.
- Bookings made within 0–30 days have the lowest cancellation rate (16.4%), making them the most reliable.
- Cancellation rates nearly double for bookings made 31–60 days in advance (31.6%).
- Guests booking more than 180 days ahead show the highest cancellation rates, reaching 40.9% for bookings made over a year in advance.
- The results indicate that early bookings are significantly more prone to cancellation, highlighting lead time as a strong predictor of cancellation risk.

```python
# 4.	What is the impact of the day of the week on booking cancellations?

# Extract arrival day
df['arrival_day'] = df['arrival_date'].dt.day_name()

# Order of weekdays
day_order = [
    'Monday', 'Tuesday', 'Wednesday',
    'Thursday', 'Friday', 'Saturday', 'Sunday'
]

# Calculate cancellation rate
cancel_rate = (
    df.groupby('arrival_day')['is_canceled']
      .mean()
      .reindex(day_order)
      .mul(100)
      .reset_index(name='Cancellation Rate (%)')
)

# Plot
plt.figure(figsize=(10,6))

ax = sns.lineplot(
    data=cancel_rate,
    x='arrival_day',
    y='Cancellation Rate (%)',
    marker='o',
    linewidth=2.8,
    color='#8E44AD'
)

for x, y in zip(cancel_rate['arrival_day'], cancel_rate['Cancellation Rate (%)']):
    plt.text(x, y + 0.25, f'{y:.1f}%', ha='center', fontsize=10)

plt.title('Cancellation Rate by Day of Arrival', fontsize=16, fontweight='bold')
plt.xlabel('Arrival Day')
plt.ylabel('Cancellation Rate (%)')

plt.grid(axis='y', linestyle='--', alpha=0.35)

plt.tight_layout()
plt.show()
```

### Key Observation

- **Friday** records the highest cancellation rate (**30.5%**), followed closely by **Saturday (30.1%)**, indicating that bookings scheduled at the beginning of the weekend are more likely to be canceled.
- **Tuesday** has the lowest cancellation rate (**24.3%**), while **Wednesday (25.5%)** also shows relatively low cancellation levels, suggesting that midweek arrivals are more reliable.
- Cancellation rates steadily increase from **Tuesday (24.3%)** to **Friday (30.5%)**, representing an increase of over **6 percentage points** before the weekend.
- After peaking on Friday, the cancellation rate declines to **26.6% on Sunday**, indicating that the highest cancellation risk is concentrated around the start of the weekend.
- The observed trend demonstrates that the day of arrival influences booking cancellations, with weekend arrivals generally experiencing higher cancellation rates than midweek arrivals.

- **Domestic guests** book an average of **64.4 days** before arrival, considerably later than guests from **Great Britain, Germany, France (74.1 days), and Italy (83.2 days)**.
- **Spanish (ESP)** guests have the shortest average lead time (**52.2 days**), even lower than domestic guests, suggesting a preference for relatively last-minute bookings.
- The lead time varies significantly across countries, demonstrating that booking behavior differs by guest origin rather than following a common international pattern.
- Major international markets generally plan their trips earlier than domestic guests, providing hotels with greater visibility into future occupancy.
